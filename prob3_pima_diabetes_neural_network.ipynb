```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8,6))

input_layer = [(1,8),(1,7),(1,6),(1,5),(1,4),(1,3),(1,2),(1,1)]

hidden_layer = [(4,7),(4,5.5),(4,4),(4,2.5),(4,1)]

#Output layer
output_layer = [(7,4.5)]

#draw input neurons
for x, y in input_layer:
    circle = plt.Circle((x,y),0.18,color="red")
    ax.add_patch(circle)

#draw hidden neurons
for x, y in hidden_layer:
    circle = plt.Circle((x,y),0.20,color="blue")
    ax.add_patch(circle)

#draw outputneuron
for x, y in output_layer:
    circle = plt.Circle((x,y),0.25,color="yellow")
    ax.add_patch(circle)

#Connect input to hidden
for ix, iy in input_layer:
    for hx, hy in hidden_layer:
        ax.plot([ix,hx],[iy,hy],color="black")

#hidden layer
for hx, hy in hidden_layer:
    ax.plot([hx,output_layer[0][0]],
            [hy,output_layer[0][1]],
            color="black")

#labels
ax.text(0.3,8.6,"Input Layer",fontsize=12)
ax.text(3.3,8.0,"Hidden Layer",fontsize=12)
ax.text(6.3,5.2,"Output Layer",fontsize=12)
ax.text(7.4,4.5,"Diabetes\nPrediction",fontsize=10)


ax.set_xlim(0,8)
ax.set_ylim(0,9)
ax.set_aspect("equal")
ax.axis("off")

plt.title("Neural Network for Pima Indians Diabetes Prediction")

plt.show()
```
